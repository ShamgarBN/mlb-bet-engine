# MLB Prediction Model — Backtest Report

Walk-forward, season-by-season backtest 2019-2025. For each season the model trains on every prior season we have data for and predicts the target season without ever seeing it.

Honest performance targets (and what we measured):

| Slice | Target | Measured |
|---|---|---|
| Moneyline — all picks | 55–58% | **60.7%** |
| Moneyline — top 10% confidence | 66–72% | **74.3%** |
| Moneyline — top 3% conviction | 70%+ | **75.9%** |
| Run line — top 10% | 70%+ | **80.7%** |
| Closing-line value (cents) | positive | **+3.48** |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../logs/backtest_v4.csv')
df
```

## Accuracy by confidence tier

```python
fig, ax = plt.subplots(figsize=(10, 5))
x = df['season'].astype(int)
ax.plot(x, df['ml_accuracy'], marker='o', label='All picks')
ax.plot(x, df['ml_top30_acc'], marker='o', label='Top 30%')
ax.plot(x, df['ml_top10_acc'], marker='o', label='Top 10%')
ax.plot(x, df['ml_top3_acc'], marker='o', label='Top 3%')
ax.axhline(0.5, color='grey', ls=':', alpha=0.5)
ax.set_xlabel('Season'); ax.set_ylabel('Accuracy'); ax.set_ylim(0.45, 0.95)
ax.set_title('Moneyline accuracy by confidence tier')
ax.legend(); ax.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

## CLV vs the closing line

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(df['season'].astype(int).astype(str), df['clv_ml'])
ax.axhline(0, color='black', lw=1)
ax.set_title('Closing-line value (cents per game, moneyline)'); ax.set_ylabel('CLV (¢)')
ax.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

## Reading the numbers

- **`ml_top3_acc` ≈ 76%** across the seven backtested seasons means: if you only bet the **top 3% of games** ranked by model confidence (≈75 picks per season), you hit ~3 out of 4. At -110 vig that is a comfortably profitable filter.
- **`clv_ml` > 0** means our model's implied probability is, on average, sharper than the closing line. Positive CLV is the strongest *forward-looking* indicator that the model has real edge.
- The OU markets are nearly perfectly efficient on public data; 52% is honest.

**This is NOT a 70%-on-all-games model.** Nothing using public data can be. It is, however, an honestly-calibrated model with a measurable edge on the moneyline and run-line markets when filtered for conviction.
